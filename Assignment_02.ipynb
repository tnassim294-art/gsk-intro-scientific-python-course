## 2. A Simulation


```python
rng = np.random.default_rng(seed = 10)
```

```python
steps = rng.choice([-1, 1], size=(5, 500))
steps
```

```python
positions = np.cumsum(steps, axis = 1)
positions
```

```python
positions.shape
```

My array has a shape of 5 rows, and 500 values in each row. Each row represents a different random walk.

## 3. Building a Plot in Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
x= np.arange(len(positions[0]))
fig1, ax = plt.subplots()
ax.plot(x,positions[0],label = 'Walk 1' , linestyle ='--')
ax.plot(x,positions[1],label = 'Walk 2',linestyle ='-' )
ax.plot(x,positions[2],label = 'Walk 3' ,linestyle ='-.')
ax.plot(x,positions[3],label = 'Walk 4',linestyle =':' )
ax.plot(x,positions[4],label = 'Walk 5')
ax.axhline(color = 'grey')
ax.set_xlabel('Steps')  
ax.set_ylabel('Position')
ax.set_title('5 Random Walks')
ax.legend()
```

## 4. Building Subplots in Matplotlib


```python
thousand_steps = rng.choice([-1, 1], size=(1000, 500))
```

```python
thousand_positions = np.cumsum(thousand_steps, axis = 1)
thousand_positions # The final position is the last element in each row
```

```python
final_values = thousand_positions[:,-1];
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2)
fig.suptitle('')
ax1.plot(x,positions[0],label = 'Walk 1' , linestyle ='--')
ax1.plot(x,positions[1],label = 'Walk 2',linestyle ='-' )
ax1.plot(x,positions[2],label = 'Walk 3' ,linestyle ='-.')
ax1.plot(x,positions[3],label = 'Walk 4',linestyle =':' )
ax1.plot(x,positions[4],label = 'Walk 5')
ax1.axhline(color = 'grey')
ax1.set_xlabel('Steps')  
ax1.set_ylabel('Position')
ax1.set_title('5 Random Walks')
ax1.legend()
ax1.set_title('5 Random Walks')

ax2.hist(final_values, bins=100)
ax2.set_title('1000 Final Positions')
ax2.set_xlabel('Final Step Value')  
ax2.set_ylabel('Occurence')

plt.tight_layout()

fig.savefig('random_walks.png')
```
